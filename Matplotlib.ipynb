# Matplotlib

Types of Data:</br>
 -> Numerical Data</br>
 -> Categorical Data

Univariate: Involves one variable (e.g., analyzing test scores).</br>

Bivariate: Involves two variables and their relationship (e.g., studying height vs. weight).</br>

Multivariate: Involves three or more variables (e.g., predicting house prices based on multiple factors).

Q.} What if values are such that One Histogram Bar is so high that we can't see others

```python
# logarithmic Scale
```

```python
# Generate exponentially increasing data
data = np.exp(np.random.normal(size=1000))  # Exponentially distributed data

# Plot the histogram
plt.hist(data, bins=50, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency (Log Scale)')
plt.title('Histogram of Exponentially Increasing Data (Log Scale)')
plt.show()
```

Solution: Use ( log = True )

```python
# Generate exponentially increasing data
data = np.exp(np.random.normal(size=1000))  # Exponentially distributed data

# Plot the histogram
plt.hist(data, bins=50, log=True, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency (Log Scale)')
plt.title('Histogram of Exponentially Increasing Data (Log Scale)')
plt.show()
```

```python
# A different way to plot graphs
batter.head()
```

```python
# 1st way
plt.figure(figsize=(12, 6))

plt.scatter(batter['avg'], batter['strike_rate'])
plt.title('Average vs Strike_rate')
plt.xlabel('Average')
plt.ylabel('Strike_rate')

plt.show()
```

```python
# 2nd way (Useful for multiple plots side by side)
fig, ax = plt.subplots(figsize=(12, 6))

ax.scatter(batter['avg'], batter['strike_rate'], color='red', marker='+')
ax.set_title('Average vs Strike_rate')
ax.set_xlabel('Average')
ax.set_ylabel('Strike_Rate')

fig.show()
```

```python
# (fig) 1st object and (ax) is the axis object
fig, ax = plt.subplots()

# passing the values of (axis) in ax
ax.plot(batter['avg'], batter['strike_rate'], marker='o')
ax.set_title('Average vs Strike_rate')
ax.set_xlabel('Average')
ax.set_ylabel('Strike_Rate')
```

```python
# subplots
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(3, 1))
```

```python
# subplots
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(12, 6))

# 1D array
# 0 -> first 
# 1 -> second
ax[0].scatter(batter['avg'], batter['strike_rate'], color='red')
ax[1].scatter(batter['avg'], batter['runs'])

ax[0].set_title('Avg vs Strike_Rate')
ax[0].set_ylabel('Strike_Rate')

ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
# subplots scatter and histogram
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 10))

ax[0, 0].scatter(batter['avg'], batter['strike_rate'], color='red')
ax[0, 1].scatter(batter['avg'], batter['runs'])
ax[1, 0].hist(batter['avg'])
ax[1, 1].hist(batter['runs'])
```

```python
# 3rd way
fig = plt.figure()

# add_subplot(No. of rows, No. of columns, index)
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'], batter['strike_rate'], color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])
```

3D Surface Plot </br>
(Used For Loss Function Plot)

Example 1

```python
# 3D Surface plots
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

xx, yy = np.meshgrid(x, y)
xx.shape, yy.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12, 8))

ax = plt.subplot(projection='3d')

P3d = ax.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(P3d)
```

Example 2

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12, 8))

ax = plt.subplot(projection='3d')

P3d = ax.plot_surface(xx, yy, z, cmap='viridis')

fig.colorbar(P3d)
```

### contour & contourf
